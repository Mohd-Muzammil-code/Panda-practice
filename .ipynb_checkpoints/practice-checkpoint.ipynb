```python
df['age'] = df['age'].fillna(df['age'].mean())
```

```python
df['salary'] = df['salary'].str.replace(",", " ", regex=False)
df['salary'] = pd.to_numeric(df['salary'], errors='coerce')
df['salary'] = df['salary'].fillna(df['salary'].median())
```

```python
df['salary'].dtype
```

```python
df.isnull().sum()
```

```python
df['city'] = df['city'].fillna(df['city'].mode()[0])
```

```python
df['department'] = df['department'].fillna(df['department'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.reset_index()
```

```python
df.duplicated(subset=["employee_id"]).sum()
```

```python
df = df.drop_duplicates(subset=["employee_id"], keep="first")
```

```python
df['joining_date'] = pd.to_datetime(df["joining_date"])
```

```python
df.sample()
```

```python
df['annual_salary'] = df['salary'] * 12
```

```python
df.head()
```

```python
def categorize(num):
    if num > 5:
        return "Exprienced candidate"
    elif num <= 5:
        return "Learner"
```

```python
df['exprience_status'] = df['experience'].apply(categorize)
```

```python
df.head()
```

```python
df.sort_values(by = "employee_id")
```

```python
df['department'] = df['department'].str.strip().str.title()
```

```python
df.sort_values(by = ['department', 'salary'],ascending=[True, False])
```

```python
df['salary_rank'] = df['salary'].rank(ascending=False)
```

```python
df.sort_values(by = "salary_rank")
```
